COPIES AND VIEWING

```python
import numpy as np
a=np.arange(10)
b=a[::2]
print(a)
print(b)
```

```python
#points to the same memory location 
print(np.shares_memory(a,b))
```

```python
b[0]=8
#this is called viewing here both variables share memory so change in one can also change in another
print(a)
print(b)
```

```python
a=np.arange(10)
#forces to create another copy
b=a[::2].copy()
print(a)
print(b)
print(np.shares_memory(a,b))
```

```python
b[0]=8
#here we separate memories so the value will not effect other
print(a)
print(b)
```
