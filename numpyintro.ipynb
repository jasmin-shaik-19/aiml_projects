creation of one dimensional array


```python
a=np.array([1,2,3,4])
print(a)
print(a.ndim)
print(a.shape)
```

creation of two dimensional array

```python
a=np.array([[1,2],[3,4]])
print(a)
print(a.ndim)
print(a.shape)
```

creation of three dimensional array

```python
a=np.array([[[1,2],[3,4]],[[5,6],[7,8]]])
print(a)
print(a.ndim)
print(a.shape)
```

functions for creating array

```python
a=np.arange(1,10,2)
print(a)
```

```python
#here linspace mean from starting point to ending point it divide them in some parts as mentioned in the argument.
a=np.linspace(0,1,6)
print(a)
```

```python
a=np.ones((3,3))
print(a)
```

```python
a=np.zeros((3,3))
print(a)
```

```python
#In diagonally we will get ones
a=np.eye(3)
print(a)
```

```python
a=np.diag([1,2,3,4])
print(a)
```

```python
print(np.diag(a))
```

BASIC DATA TYPES

```python
a=np.arange(10)
print(a.dtype)
```

```python
a=np.arange(10,dtype="float64")
print(a)
print(a.dtype)
```

OTHER DATATYPE

```python
a=np.array([1+2j,2+4j])
print(a.dtype)
```

```python
a=np.array([True,False])
print(a)
print(a.dtype)
```

```python
a=np.array(["jasmin","shaik"])
print(a)
print(a.dtype)
#u means unicode string type and 6 means highest length
```
